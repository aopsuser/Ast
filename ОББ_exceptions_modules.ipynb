# Python: ОББ — Мұрагерлік, Ерекше жағдайларды өңдеу (try-except), Модульдер мен пакеттер


Бұл ноутбук Python тіліндегі үш маңызды тақырыпты қамтиды: ОББ-дағы мұрагерлік механизмі, try-except арқылы ерекше жағдайларды өңдеу, және модульдер мен пакеттерді пайдалану. Әрбір бөлімде теориялық түсіндірме, код мысалдары және практикалық тапсырмалар берілген.


**Мақсат:** Мұрагерлік механизмін түсіну және іс жүзінде қолдана білу.

**Теория:**  
Мұрагерлік (inheritance) бір класстың (бала класс) басқа класстың (ата-ана класс) атрибуттары мен әдістерін мұраға алу мүмкіндігі. Бұл кодты қайта пайдалануды жеңілдетеді және бағдарламаның логикалық құрылымын жақсартады. Python-да мұрагерлік `class БалаКласс(АтаАнаКласс):` синтаксисімен жазылады. `super()` функциясы арқылы ата-ана класстың конструкторы мен әдістеріне қол жеткізуге болады.

```python
# Ата-ана класс — барлық жануарлар үшін жалпы базалық класс
class Animal:
    def __init__(self, name, age):
        self.name = name  # жануардың аты
        self.age = age    # жануардың жасы

    def speak(self):
        # Бұл әдіс бала кластарда қайта анықталуы тиіс
        return "..."

    def info(self):
        # Жануар туралы жалпы ақпаратты шығару
        return f"Аты: {self.name}, Жасы: {self.age}"


# Бала класс — Animal класынан мұрагер
class Dog(Animal):
    def __init__(self, name, age, breed):
        super().__init__(name, age)  # ата-ана класстың конструкторын шақыру
        self.breed = breed           # иттің тұқымы

    def speak(self):
        # speak әдісін қайта анықтау (override)
        return "Арп-арп!"

    def fetch(self):
        # Итке ғана тән арнайы әдіс
        return f"{self.name} допты алып келді!"


# Тағы бір бала класс
class Cat(Animal):
    def __init__(self, name, age, indoor):
        super().__init__(name, age)  # ата-ана класстың конструкторын шақыру
        self.indoor = indoor         # үйде ұсталатын мысық па?

    def speak(self):
        # speak әдісін қайта анықтау
        return "Мяу!"


# Объектілер жасау және тестілеу
dog = Dog("Барсик", 3, "Немістік шепард")
cat = Cat("Мурзик", 5, True)

print(dog.info())    # ата-ана класстан мұраға алынған әдіс
print(dog.speak())   # бала класта қайта анықталған әдіс
print(dog.fetch())   # бала класқа ғана тән әдіс
print(cat.info())    # ата-ана класстан мұраға алынған әдіс
print(cat.speak())   # бала класта қайта анықталған әдіс
```

### 1.1 Көп деңгейлі мұрагерлік (Multilevel Inheritance)

**Теория:**  
Көп деңгейлі мұрагерліктe бала класс өз кезегінде басқа кластың ата-ана класы бола алады. Мысалы, `A → B → C` тізбегінде `C` класы `B` арқылы `A` класының барлық мүмкіндіктерін иеленеді. Бұл иерархиялық модельдерді (мысалы, биологиялық классификация, компания құрылымы) кодта бейнелеуге өте қолайлы. `isinstance()` функциясы объектінің белгілі бір кластың данасы екенін тексереді.

```python
# Базалық класс — тіршілік иесі
class LivingBeing:
    def __init__(self, name):
        self.name = name  # тіршілік иесінің аты

    def breathe(self):
        # Барлық тіршілік иелеріне тән қасиет
        return f"{self.name} тыныс алады"


# Екінші деңгей — LivingBeing-тен мұрагер
class Mammal(LivingBeing):
    def __init__(self, name, warm_blooded=True):
        super().__init__(name)          # LivingBeing конструкторын шақыру
        self.warm_blooded = warm_blooded  # жылықанды ма?

    def feed_young(self):
        # Сүтқоректілерге тән қасиет
        return f"{self.name} балапанын сүтпен қоректендіреді"


# Үшінші деңгей — Mammal-дан мұрагер
class Human(Mammal):
    def __init__(self, name, profession):
        super().__init__(name)          # Mammal конструкторын шақыру
        self.profession = profession    # адамның мамандығы

    def work(self):
        # Адамға ғана тән әдіс
        return f"{self.name} — {self.profession}, жұмыс жасайды"


# Объект жасап, барлық деңгейдің әдістерін тексеру
person = Human("Айдар", "бағдарламашы")

print(person.breathe())     # LivingBeing-тен мұраға алынған
print(person.feed_young())  # Mammal-дан мұраға алынған
print(person.work())        # Human-ның өз әдісі

# isinstance арқылы мұрагерлік байланысын тексеру
print(isinstance(person, Human))       # True — тікелей класс
print(isinstance(person, Mammal))      # True — ата-ана класс
print(isinstance(person, LivingBeing)) # True — баба класс
```

### 1.2 Полиморфизм (Polymorphism)

**Теория:**  
Полиморфизм бір интерфейстің (функция немесе әдіс атауы) әртүрлі кластарда әртүрлі жүзеге асырылуы. Python-да полиморфизм динамикалық түрде жұмыс істейді: бір атаулы әдіс әртүрлі объектілерде әртүрлі нәтиже беруі мүмкін. Бұл кодты икемді және кеңейтілетін етеді, `if-else` тізбектерінің санын азайтады.

```python
# Геометриялық фигуралар арқылы полиморфизм мысалы
class Shape:
    def area(self):
        # Ауданды есептейтін базалық әдіс
        raise NotImplementedError("Бала класс area() әдісін іске асыруы тиіс")

    def describe(self):
        # Фигура сипаттамасы
        return f"Бұл {self.__class__.__name__}, ауданы = {self.area():.2f}"


class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius  # шеңбер радиусы

    def area(self):
        # Шеңбер ауданын есептеу формуласы
        import math
        return math.pi * self.radius ** 2


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width    # тіктөртбұрыш ені
        self.height = height  # тіктөртбұрыш биіктігі

    def area(self):
        # Тіктөртбұрыш ауданын есептеу
        return self.width * self.height


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base      # үшбұрыш негізі
        self.height = height  # үшбұрыш биіктігі

    def area(self):
        # Үшбұрыш ауданын есептеу
        return 0.5 * self.base * self.height


# Полиморфизмді тексеру — бір функция, әртүрлі объектілер
shapes = [Circle(5), Rectangle(4, 6), Triangle(3, 8)]

for shape in shapes:
    # describe() әдісі әр кластa әртүрлі area() шақырады
    print(shape.describe())
```

## 2-бөлім: Ерекше жағдайларды өңдеу (Exception Handling)

**Мақсат:** try-except блоктарын пайдаланып, бағдарлама қателерін дұрыс өңдеуді үйрену.

**Теория:**  
Ерекше жағдайлар (exceptions) бағдарлама орындалу кезінде туындайтын қателер. Python-да `try` блогына қате шығуы мүмкін код орналасады, `except` блогы қатені ұстап алып, өңдейді. `else` блогы қате болмаған жағдайда орындалады, `finally` блогы қате болса да, болмаса да әрдайым орындалады. Бұл механизм бағдарламаның кенеттен тоқтап қалуының алдын алады.

```python
10/0
```

```python
# Негізгі try-except синтаксисін үйрену

# Мысал 1: ZeroDivisionError — нөлге бөлу қатесі
def safe_divide(a, b):
    try:
        result = a / b  # нөлге бөлуге тырысу
    except ZeroDivisionError:
        # Нөлге бөлу қатесін өңдеу
        print("Қате: нөлге бөлуге болмайды!")
        return None
    else:
        # Қате болмаса орындалады
        print(f"Нәтиже: {a} / {b} = {result}")
        return result
    finally:
        # Қате болса да, болмаса да орындалады
        print("safe_divide функциясы аяқталды")


print("--- Бірінші шақыру ---")
safe_divide(10, 2)   # қатесіз жағдай

print("\n--- Екінші шақыру ---")
safe_divide(10, 0)   # қате шығатын жағдай
```

```python
# Мысал 2: Бірнеше қате түрін өңдеу
def process_input(data):
    try:
        # Санды áрі бүтін санға айналдырып, 10-ға бөлу
        number = int(data)
        result = 100 / number
        return result
    except ValueError:
        # Санға айналдыру мүмкін болмаса
        print(f"Қате: '{data}' саны дұрыс емес формат")
    except ZeroDivisionError:
        # Нөлге бөлу жағдайы
        print("Қате: нөл енгізілді, нөлге бөлуге болмайды")
    except Exception as e:
        # Күтпеген кез-келген басқа қателерді ұстау
        print(f"Белгісіз қате: {type(e).__name__}: {e}")


# Әртүрлі деректермен тестілеу
test_inputs = ["5", "0", "бес", "3.14", "20"]

for item in test_inputs:
    print(f"Енгізілген: {item!r} → ", end="")
    result = process_input(item)
    if result is not None:
        # Нәтиже дұрыс болса шығару
        print(f"Нәтиже = {result:.2f}")
```

### 2.1 Өзіндік ерекше жағдайлар жасау (Custom Exceptions)

**Теория:**  
Python-да `Exception` класынан мұрагерлік арқылы өзіндік ерекше жағдай кластарын жасауға болады. Бұл үлкен жобаларда қателерді нақты категориялауға мүмкіндік береді. Өзіндік ерекше жағдайлар арқылы кодты оқуға ыңғайлы, мәнді қате хабарламалары жазуға болады. `raise` кілт сөзі арқылы ерекше жағдай қолмен шығарылады.

```python
# Банк есебі үшін өзіндік ерекше жағдайлар

class BankError(Exception):
    """Банк операцияларының базалық ерекше жағдай класы"""
    pass


class InsufficientFundsError(BankError):
    """Шотта жеткілікті қаражат болмаған жағдайда шығарылатын қате"""
    def __init__(self, balance, amount):
        self.balance = balance  # ағымдағы баланс
        self.amount = amount    # талап етілген сома
        super().__init__(
            f"Жеткіліксіз қаражат: баланс {balance}₸, талап {amount}₸"
        )


class NegativeAmountError(BankError):
    """Теріс сома енгізілгенде шығарылатын қате"""
    def __init__(self, amount):
        self.amount = amount  # енгізілген теріс сома
        super().__init__(f"Теріс сома енгізуге болмайды: {amount}₸")


class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner      # шот иесінің аты
        self.balance = balance  # бастапқы баланс

    def deposit(self, amount):
        # Шотқа ақша салу операциясы
        if amount <= 0:
            raise NegativeAmountError(amount)  # теріс сома қатесін шығару
        self.balance += amount
        return f"{amount}₸ салынды. Баланс: {self.balance}₸"

    def withdraw(self, amount):
        # Шоттан ақша алу операциясы
        if amount <= 0:
            raise NegativeAmountError(amount)  # теріс сома қатесін шығару
        if amount > self.balance:
            raise InsufficientFundsError(self.balance, amount)  # жеткіліксіз қаражат
        self.balance -= amount
        return f"{amount}₸ алынды. Баланс: {self.balance}₸"


# Банк шотын тестілеу
account = BankAccount("Зарина", 5000)

operations = [
    ("deposit", 2000),   # дұрыс операция
    ("withdraw", 3000),  # дұрыс операция
    ("withdraw", 10000), # жеткіліксіз қаражат
    ("deposit", -500),   # теріс сома
]

for op_type, amount in operations:
    try:
        if op_type == "deposit":
            print(account.deposit(amount))
        else:
            print(account.withdraw(amount))
    except InsufficientFundsError as e:
        # Жеткіліксіз қаражат қатесін өңдеу
        print(f"⚠ Қате: {e}")
    except NegativeAmountError as e:
        # Теріс сома қатесін өңдеу
        print(f"⚠ Қате: {e}")
```

### 2.2 Контекст менеджері (Context Manager) — with операторы

**Теория:**  
Контекст менеджері (`with` операторы) ресурстарды (файлдар, желі қосылымдары) автоматты түрде ашып-жабуды қамтамасыз етеді. `__enter__` және `__exit__` арнайы әдістерін іске асыру арқылы өзіндік контекст менеджерлерін жасауға болады. Бұл `finally` блоктарын жазудың қысқа және оқылымды балама жолы.

```python
# Контекст менеджері арқылы ресурстарды басқару

class DatabaseConnection:
    """Деректер қоры қосылымын имитациялайтын класс"""

    def __init__(self, db_name):
        self.db_name = db_name  # деректер қорының атауы
        self.connected = False  # қосылым күйі

    def __enter__(self):
        # with блогы басталғанда автоматты шақырылады
        self.connected = True
        print(f"✓ {self.db_name} деректер қорына қосылды")
        return self  # with блогындағы as мәні

    def __exit__(self, exc_type, exc_val, exc_tb):
        # with блогы аяқталғанда автоматты шақырылады
        self.connected = False
        print(f"✓ {self.db_name} деректер қорынан ажырасылды")
        if exc_type is not None:
            # Қате болса, оны жазып өту
            print(f"⚠ Қате болды: {exc_type.__name__}: {exc_val}")
        return False  # қатені жоймай, одан әрі жіберу

    def query(self, sql):
        # SQL сұрау орындау
        if not self.connected:
            raise ConnectionError("Деректер қорына қосылым жоқ")
        return f"SQL сұрауы орындалды: {sql}"


# Контекст менеджерін тестілеу
print("--- Қалыпты операция ---")
with DatabaseConnection("students_db") as db:
    # Блок ішінде қосылым автоматты ашылды
    result = db.query("SELECT * FROM students")
    print(result)
# Блоктан шыққан соң қосылым автоматты жабылды

print()
print("--- Қате болған жағдай ---")
try:
    with DatabaseConnection("grades_db") as db:
        result = db.query("SELECT * FROM grades")
        print(result)
        # Арнайы қате шығару
        raise ValueError("SQL синтаксис қатесі")
except ValueError as e:
    # Қате жоғарыға жіберілді
    print(f"Жоғары деңгейде өңделді: {e}")
```

## 3-бөлім: Модульдер мен пакеттер (Modules and Packages)

**Мақсат:** Python-дағы модульдер мен пакеттердің жұмысын түсіну және стандартты кітапхананы тиімді қолдану.

**Теория:**  
Модуль — бір `.py` кеңейтімді файл, ал пакет — `__init__.py` файлы бар қалта. `import` операторы арқылы модульдер жүктеледі. Python-ның стандартты кітапханасы (Standard Library) жүздеген пайдалы модульдерді қамтиды: `math`, `os`, `sys`, `datetime`, `random`, `collections` т.б. ### 3.1 Модуль жасау және импорттау

**Теория:**  
Өзіндік модуль жасау үшін функциялар мен кластарды бөлек `.py` файлға жазу жеткілікті. Модульдер `import module_name`, `from module_name import function_name`, немесе `import module_name as alias` түрінде импортталады. `if __name__ == '__main__':` конструкциясы файл тікелей іске қосылғанда ғана орындалатын кодты анықтайды. Бұл кодты қайта пайдалануды жеңілдетеді.

```python
# Модуль ретінде пайдаланылатын код жасау
# Бұл ұяшық utils.py модулінің мазмұнын имитациялайды

# ---- utils.py мазмұны ----

def celsius_to_fahrenheit(celsius):
    """Цельсийді Фаренгейтке айналдыру"""
    # Айналдыру формуласы: F = C * 9/5 + 32
    return celsius * 9 / 5 + 32


def fahrenheit_to_celsius(fahrenheit):
    """Фаренгейтті Цельсийге айналдыру"""
    # Кері айналдыру формуласы
    return (fahrenheit - 32) * 5 / 9


def is_palindrome(text):
    """Мәтіннің палиндром екенін тексеру"""
    # Бос орындар мен регистрді ескермей тексеру
    cleaned = text.lower().replace(" ", "")
    return cleaned == cleaned[::-1]  # кері тізбекпен салыстыру


def flatten_list(nested):
    """Кірістірілген тізімді жалпақтау (бір деңгей)"""
    result = []
    for item in nested:
        if isinstance(item, list):
            result.extend(item)  # тізімді жайып қосу
        else:
            result.append(item)  # элементті тікелей қосу
    return result


def word_frequency(sentence):
    """Сөйлемдегі сөздердің жиілігін есептеу"""
    words = sentence.lower().split()  # кіші әріпке айналдырып бөлу
    frequency = {}
    for word in words:
        frequency[word] = frequency.get(word, 0) + 1  # санауышты арттыру
    return dict(sorted(frequency.items(), key=lambda x: x[1], reverse=True))


# ---- Функцияларды тестілеу ----
print("Температура айналдыру:")
print(f"100°C = {celsius_to_fahrenheit(100)}°F")
print(f"212°F = {fahrenheit_to_celsius(212):.1f}°C")

print("\nПалиндром тексеру:")
test_words = ["level", "python", "madam", "racecar", "Алма"]
for word in test_words:
    # Palindrom болса True, болмаса False шығарылады
    status = "Палиндром ✓" if is_palindrome(word) else "Палиндром емес ✗"
    print(f"  '{word}': {status}")

print("\nТізімді жайлау:")
nested = [[1, 2, 3], [4, 5], [6, 7, 8, 9]]
print(f"  Бастапқы: {nested}")
print(f"  Жайылған: {flatten_list(nested)}")
```

## 4-бөлім: Кешенді практикалық жоба

**Мақсат:** ОББ мұрагерлігін, ерекше жағдайларды өңдеуді және модульдерді бір жобада біріктіру.

**Теория:**  
Нақты жобаларда ОББ, қателерді өңдеу және стандартты кітапхана бір-бірімен тығыз байланыста қолданылады. Бұл бөлімде студент тіркеу жүйесі жасалады: класс иерархиясы мұрагерлік арқылы, деректерді тексеру ерекше жағдайлар арқылы, ал уақыт пен кездейсоқ деректер стандартты модульдер арқылы басқарылады.

```python
# Студент тіркеу жүйесі — кешенді мысал
import random
from datetime import datetime
from collections import defaultdict


# --- Өзіндік ерекше жағдайлар ---
class RegistrationError(Exception):
    """Тіркеу жүйесінің базалық қате класы"""
    pass


class DuplicateStudentError(RegistrationError):
    """Студент бұрын тіркелген жағдайдағы қате"""
    pass


class InvalidGradeError(RegistrationError):
    """Жарамсыз баға енгізілгендегі қате"""
    pass


# --- Класс иерархиясы ---
class Person:
    """Базалық адам класы"""
    def __init__(self, first_name, last_name, birth_year):
        self.first_name = first_name    # аты
        self.last_name = last_name      # тегі
        self.birth_year = birth_year    # туған жылы

    @property
    def full_name(self):
        # Толық атын қасиет ретінде қайтару
        return f"{self.last_name} {self.first_name}"

    @property
    def age(self):
        # Жасты автоматты есептеу
        return datetime.now().year - self.birth_year

    def __str__(self):
        # Объектіні жолға айналдыру
        return f"{self.full_name} ({self.age} жас)"


class Student(Person):
    """Студент класы — Person-нан мұрагер"""
    def __init__(self, first_name, last_name, birth_year, student_id):
        super().__init__(first_name, last_name, birth_year)  # Person конструкторы
        self.student_id = student_id  # студент куәлігі нөмірі
        self.grades = {}              # пәндер бойынша бағалар сөздігі
        self.enrolled_date = datetime.now()  # тіркелу күні

    def add_grade(self, subject, grade):
        # Баға қосу — дұрыстығын тексеру
        if not (0 <= grade <= 100):
            raise InvalidGradeError(
                f"Жарамсыз баға: {grade}. Баға 0 мен 100 арасында болуы тиіс"
            )
        self.grades[subject] = grade

    @property
    def gpa(self):
        # Орташа баллды есептеу
        if not self.grades:
            return 0.0  # бағалар жоқ болса нөл
        return sum(self.grades.values()) / len(self.grades)

    def get_letter_grade(self):
        # Балл бойынша әріп бағасын анықтау
        gpa = self.gpa
        if gpa >= 90:
            return 'A'  # үздік
        elif gpa >= 75:
            return 'B'  # жақсы
        elif gpa >= 60:
            return 'C'  # қанағаттанарлық
        else:
            return 'F'  # қанағаттанарлықсыз

    def __str__(self):
        # Студент туралы толық ақпарат
        return (
            f"Студент: {self.full_name} | "
            f"ID: {self.student_id} | "
            f"GPA: {self.gpa:.1f} ({self.get_letter_grade()})"
        )


class GraduateStudent(Student):
    """Магистрант класы — Student-тен мұрагер"""
    def __init__(self, first_name, last_name, birth_year, student_id, research_topic):
        super().__init__(first_name, last_name, birth_year, student_id)  # Student конструкторы
        self.research_topic = research_topic  # ғылыми зерттеу тақырыбы
        self.publications = []                # жарияланымдар тізімі

    def add_publication(self, title, year):
        # Жарияланым қосу
        self.publications.append({"title": title, "year": year})

    def __str__(self):
        # Магистрант туралы толық ақпарат
        base = super().__str__()  # Student-тің __str__ әдісін алу
        return f"{base} | Тақырып: {self.research_topic}"


# --- Тіркеу жүйесі ---
class RegistrationSystem:
    """Студент тіркеу жүйесін басқаратын класс"""

    def __init__(self):
        self.students = {}              # id:student жұбы
        self.departments = defaultdict(list)  # факультеттер бойынша студенттер

    def register(self, student, department):
        # Жаңа студентті тіркеу
        if student.student_id in self.students:
            raise DuplicateStudentError(
                f"Студент {student.student_id} бұрын тіркелген!"
            )
        self.students[student.student_id] = student  # сөздікке қосу
        self.departments[department].append(student)  # факультетке қосу
        return f"✓ {student.full_name} тіркелді ({department})"

    def get_top_students(self, n=3):
        # GPA бойынша үздік n студентті табу
        sorted_students = sorted(
            self.students.values(),
            key=lambda s: s.gpa,  # GPA бойынша сұрыптау кілті
            reverse=True           # кему ретімен
        )
        return sorted_students[:n]

    def department_stats(self):
        # Факультеттер бойынша статистика шығару
        stats = {}
        for dept, students in self.departments.items():
            gpas = [s.gpa for s in students if s.grades]  # бағасы бар студенттер
            stats[dept] = {
                "count": len(students),             # студент саны
                "avg_gpa": sum(gpas) / len(gpas) if gpas else 0  # орташа GPA
            }
        return stats


# --- Тестілеу ---
print("="*60)
print("СТУДЕНТ ТІРКЕУ ЖҮЙЕСІ")
print("="*60)

# Тіркеу жүйесін жасау
system = RegistrationSystem()

# Студенттер деректері
students_info = [
    (Student("Айгерім", "Сейтқали", 2002, "STU001"), "CS"),
    (Student("Берік", "Асанов", 2001, "STU002"), "Math"),
    (Student("Гүлнар", "Ибраева", 2003, "STU003"), "CS"),
    (GraduateStudent("Дамир", "Нұрланов", 1998, "GRAD001", "Машиналық оқыту"), "CS"),
]

# Студенттерді тіркеу
print("\n--- Тіркеу ---")
for student, dept in students_info:
    try:
        msg = system.register(student, dept)
        print(msg)
    except DuplicateStudentError as e:
        # Қайталанған тіркеу қатесін өңдеу
        print(f"⚠ {e}")

# Бағалар қосу
print("\n--- Бағалар ---")
grades_data = [
    ("STU001", [("Python", 92), ("Математика", 88), ("Алгоритмдер", 95)]),
    ("STU002", [("Математика", 97), ("Статистика", 94)]),
    ("STU003", [("Python", 78), ("Математика", 72), ("Алгоритмдер", 150)]),  # қате баға
    ("GRAD001", [("Machine Learning", 96), ("Deep Learning", 94)]),
]

for student_id, grades in grades_data:
    student = system.students[student_id]
    for subject, grade in grades:
        try:
            student.add_grade(subject, grade)
            print(f"  {student.first_name}: {subject} → {grade}")
        except InvalidGradeError as e:
            # Жарамсыз баға қатесін өңдеу
            print(f"  ⚠ {student.first_name}: {e}")

# Үздік студенттер
print("\n--- Үздік студенттер (TOP 3) ---")
for i, s in enumerate(system.get_top_students(3), 1):
    # Орын нөмірімен шығару
    print(f"  {i}. {s}")

# Факультет статистикасы
print("\n--- Факультет статистикасы ---")
for dept, stat in system.department_stats().items():
    # Орташа GPA-мен шығару
    print(f"  {dept}: {stat['count']} студент, орт. GPA = {stat['avg_gpa']:.1f}")
```

## 5-бөлім: Практикалық тапсырмалар

**Мақсат:** Үйренген материалдарды өз бетінше қолдана білу дағдысын дамыту.

Төмендегі тапсырмаларды орындаңыз. Әр тапсырма осы ноутбуктің тиісті бөлімдеріндегі материалдарды практикада қолдануға арналған.

```python
# Тапсырма 1: Мұрагерлік
# Транспорт құралдары иерархиясын жасаңыз:
# - Vehicle (базалық): brand, year, speed
# - Car(Vehicle): doors, fuel_type
# - ElectricCar(Car): battery_capacity, range_km
# ElectricCar объектісін жасап, барлық атрибуттары мен
# мұраға алынған әдістерін шығарыңыз.

# --- Кодыңызды осында жазыңыз ---
class Vehicle:
    def __init__(self, brand, year, speed):
        self.brand = brand
        self.year = year
        self.speed = speed

    def get_info(self):
        return f"Brand: {self.brand}, Year: {self.year}, Speed: {self.speed} km/h"

class Car(Vehicle):
    def __init__(self, brand, year, speed, doors, fuel_type):
        super().__init__(brand, year, speed)
        self.doors = doors
        self.fuel_type = fuel_type

class ElectricCar(Car):
    def __init__(self, brand, year, speed, doors, fuel_type, battery_capacity, range_km):
        super().__init__(brand, year, speed, doors, fuel_type)
        self.battery_capacity = battery_capacity
        self.range_km = range_km

    def show_all_specs(self):
        base_info = self.get_info()
        return (f"{base_info}, Doors: {self.doors}, Fuel: {self.fuel_type}, "
                f"Battery: {self.battery_capacity} kWh, Range: {self.range_km} km")

my_ev = ElectricCar("Tesla", 2023, 250, 4, "Electric", 100, 600)

print(my_ev.show_all_specs())
```

```python
# Тапсырма 2: Ерекше жағдайлар
# Температураны өлшейтін функция жасаңыз:
# - Абсолют нөлден (-273.15°C) төмен мән болса → CustomTemperatureError
# - Сан емес мән берілсе → TypeError өңдеу
# - Дұрыс мән болса → Кельвинге айналдырып қайтару
# Кемінде 4 түрлі енгізу мәнімен тестілеңіз.

# --- Кодыңызды осында жазыңыз ---
class CustomTemperatureError(Exception):
    pass

def measure_temperature(celsius):
    try:
        if not isinstance(celsius, (int, float)):
            raise TypeError("Қате: Температура сан түрінде болуы керек!")

        if celsius < -273.15:
            raise CustomTemperatureError(f"Қате: {celsius}°C абсолют нөлден төмен!")

        kelvin = celsius + 273.15
        return f"{celsius}°C = {kelvin} K"

    except (TypeError, CustomTemperatureError) as e:
        return str(e)

test_values = [25, -300, "25", 0]

for value in test_values:
    print(f"Енгізілген мән: {value} -> Нәтиже: {measure_temperature(value)}")
```

```python
# Тапсырма 3: Модульдер
# datetime, random, collections модульдерін пайдаланып:
# 1) Кездейсоқ 20 студент баллдарын генерациялаңыз (50-100 аралығы)
# 2) Баллдарды А(90+), B(75-89), C(60-74), F(<60) категорияларға бөліңіз
# 3) Counter арқылы әр категориядан неше студент барын шығарыңыз
# 4) Ең жоғары және ең төмен балл алған уақытты datetime арқылы жазыңыз

# --- Кодыңызды осында жазыңыз ---
import random
from collections import Counter
from datetime import datetime

scores = [random.randint(50, 100) for _ in range(20)]

categories = []
for score in scores:
    if score >= 90:
        categories.append('A')
    elif score >= 75:
        categories.append('B')
    elif score >= 60:
        categories.append('C')
    else:
        categories.append('F')

counts = Counter(categories)

max_score = max(scores)
min_score = min(scores)
current_time = datetime.now().strftime("%Y-%m-%d %H:%M:%S")

print(f"Барлық баллдар: {scores}")
print(f"Категориялар саны: {dict(counts)}")
print(f"Ең жоғары балл: {max_score} (тіркелген уақыты: {current_time})")
print(f"Ең төмен балл: {min_score} (тіркелген уақыты: {current_time})")
```
